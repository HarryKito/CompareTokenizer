```python
text = """
この件はお持ち帰りして検討させていただいてよろしいでしょうか。
即答できなく恐縮ですが
一応内部でまとめますので申し訳ございませんが後程またよろしくお願いいたします。
"""
```

```python
import MeCab
import pandas as pd
from word2word import Word2word
from posmap import jp_ko_pos_map
```

```python
wakati = MeCab.Tagger()
MeCab_data = wakati.parse(text)
# Translator
ja2ko = Word2word("ja", "ko")
```

```python
rows=[]

for line in MeCab_data.splitlines():
    if '\t' not in line: # character delimiter = tab
        continue

    surface, features = line.split('\t', 1)
    feature_list = features.split(',')

    # base_form for translate (or use the surface)
    base_form = feature_list[6] if feature_list[6] != "*" else surface

    try:
        ko_words = ja2ko(base_form)[0]
        kr = ", ".join(ko_words) if ko_words else None
    except Exception:
        kr = None

    rows.append([surface] + feature_list + [kr])
```

```python
columns = [
    "Text",
    "pos1", # Part Of Speeches
    "pos2",
    "pos3",
    "pos4",
    "conj_type",
    "conj_form",
    "base_form",
    "Reading",
    "pronunciation",
    "Kr"
]

df = pd.DataFrame(rows, columns=columns)
df.head()
```
